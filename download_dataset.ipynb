```python
# Download the dataset
import pandas as pd

url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/we_fn_use_c_marketing_customer_value_analysis.csv"
df = pd.read_csv(url)

display(df)
```

```python
# Clean the column names
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Renalame specific columns
df = df.rename(columns={'employmentstatus': 'employment_status',
                        'customer': 'customer_id'})

# Convert date columns to datetime format
df['effective_to_date'] = pd.to_datetime(df['effective_to_date'])

# Convert date columns values to "dd-mm-yyyy" format
df['effective_to_date'] = df['effective_to_date'].dt.strftime('%d-%m-%Y')

df
```

```python
# Save the dataset to a CSV file
df.to_csv("marketing_customer_value_analysis.csv", index=False)
```
